## Task Description
Tag each word in a sentence with its grammatical part of speech (Noun, Verb, Adjective, etc.).

1. Install necessary libraries.
2. Load a real dataset or sample data.
3. Load a pre-trained model suitable for this task.
4. Run inference and analyze the results in depth.

## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
text = "I will quickly run to the new store."
print(f"Input: {text}")
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
# We use a model fine-tuned for POS tagging
pos_tagger = pipeline("token-classification", model="vblagoje/bert-english-uncased-finetuned-pos", aggregation_strategy="simple")
print("POS Tagger loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
tags = pos_tagger(text)

print("Tags:")
for tag in tags:
    print(f"Word: {tag['word']} -> POS: {tag['entity_group']} (Conf: {tag['score']:.2f})")
```
