## Step 2: Setup and Imports

```python
from transformers import pipeline
```

## Step 3: Run Inference

```python
# We use a token classification model fine-tuned for POS
pos = pipeline("token-classification", model="vblagoje/bert-english-uncased-finetuned-pos", aggregation_strategy="simple")
text = "I ran quickly to the store."
tags = pos(text)

for tag in tags:
    print(f"Word: {tag['word']}, POS: {tag['entity_group']}")
```
