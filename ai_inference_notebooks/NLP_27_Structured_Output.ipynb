## 2. Imports
Import libraries.

```
import outlines
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
```

## 3. Load Data
Load dataset or input data.

```
prompt = "Generate a character profile for a RPG game."
print(f"Prompt: {prompt}")
```

## 4. Data Cleaning
Preprocess the data.

```
# Define the schema
schema = '''
{
  "title": "Character",
  "type": "object",
  "properties": {
    "name": { "type": "string" },
    "class": { "type": "string", "enum": ["Warrior", "Mage", "Thief"] },
    "level": { "type": "integer" }
  },
  "required": ["name", "class", "level"]
}
'''
print("Schema defined.")
```

## 5. Load Model
Load the model from Hugging Face.

```
model_id = "HuggingFaceTB/SmolLM2-1.7B-Instruct"
model = outlines.models.transformers(model_id, device="cuda" if torch.cuda.is_available() else "cpu")
print(f"Loaded {model_id} with Outlines.")
```

## 6. Inference
Run the model.

```
# Generate structured output
generator = outlines.generate.json(model, schema)
result = generator(prompt)
print("Generated Result:")
print(result)
results = str(result) # For Kitchen Sink
```
