# Notebook 3: Computer Vision

In this notebook, we will explore **Computer Vision (CV)**. This field enables computers to "see" and interpret images.

We will continue using the `pipeline` function.

## Tasks we will cover:
1. **Image Classification**: Identifying what is in an image.
2. **Object Detection**: Identifying multiple objects and their positions in an image.

## Setup
First, we need to download some sample images to test our models.

```python
import requests
from PIL import Image
from transformers import pipeline

# Helper function to download an image
def download_image(url):
    return Image.open(requests.get(url, stream=True).raw)

# Download a sample image of a cat
image_url = "http://images.cocodataset.org/val2017/000000039769.jpg"
image = download_image(image_url)

# Display the image
# In Jupyter, simply typing the variable name of an image displays it
image
```

## Task 1: Image Classification

Image classification assigns a label to the whole image. For example, "cat", "dog", "car".

We will use the `"image-classification"` pipeline.

```python
# Create an image classification pipeline
classifier = pipeline("image-classification")

# Classify the image we downloaded
results = classifier(image)

# Print the top predictions
for result in results:
    print(f"Label: {result['label']}, Score: {round(result['score'], 4)}")
```

## Task 2: Object Detection

Object detection is more advanced. It finds objects in the image and draws a "bounding box" around them.

We will use the `"object-detection"` pipeline.

```python
# Create an object detection pipeline
detector = pipeline("object-detection")

# Detect objects in the image
detections = detector(image)

print(f"Found {len(detections)} objects.")
for detection in detections:
    print(detection)
```

### Visualizing the Detection
The output gives us coordinates (boxes), but it's hard to visualize just by reading numbers. Let's draw boxes on the image using Python code.

```python
from PIL import ImageDraw

# Create a copy of the image so we don't modify the original
image_with_boxes = image.copy()
draw = ImageDraw.Draw(image_with_boxes)

for detection in detections:
    box = detection['box']
    label = detection['label']
    score = detection['score']
    
    # Draw the bounding box
    # The box dictionary usually has xmin, ymin, xmax, ymax
    draw.rectangle([box['xmin'], box['ymin'], box['xmax'], box['ymax']], outline="red", width=3)
    
    # Draw the text label
    draw.text((box['xmin'], box['ymin']), f"{label} {round(score, 2)}", fill="red")

# Display the image with boxes
image_with_boxes
```
