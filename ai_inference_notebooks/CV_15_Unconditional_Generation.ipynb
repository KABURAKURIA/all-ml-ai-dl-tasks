## 2. Imports
Importing the necessary Python modules.

```python
from diffusers import DDPMPipeline
import torch
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
print("Generating a random face image...")
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
# Load DDPM trained on CelebA-HQ
pipe = DDPMPipeline.from_pretrained("google/ddpm-celebahq-256")
pipe.to("cuda" if torch.cuda.is_available() else "cpu")
print("DDPM loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
image = pipe(num_inference_steps=25).images[0]
display(image)
```
