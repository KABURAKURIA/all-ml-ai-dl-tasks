## 2. Imports
Import libraries.

```
import cv2
import numpy as np
from PIL import Image
from diffusers import StableDiffusionControlNetPipeline, ControlNetModel, UniPCMultistepScheduler
import torch
from controlnet_aux import CannyDetector
```

## 3. Load Data
Load dataset or input data.

```
# Create a simple dummy image with a shape if no input
# Or download one
import requests
from io import BytesIO
url = "https://hf.co/datasets/huggingface/documentation-images/resolve/main/diffusers/input_image_vermeer.png"
response = requests.get(url)
init_image = Image.open(BytesIO(response.content)).convert("RGB").resize((512, 512))
# init_image.show() # Can't show in script env, but good for nb
print("Loaded image.")
```

## 4. Data Cleaning
Preprocess the data.

```
# Get Canny Edge
canny = CannyDetector()
canny_image = canny(init_image)
# canny_image.show()
print("Edges detected.")
```

## 5. Load Model
Load the model.

```
controlnet = ControlNetModel.from_pretrained("lllyasviel/sd-controlnet-canny", torch_dtype=torch.float16)
pipe = StableDiffusionControlNetPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", controlnet=controlnet, torch_dtype=torch.float16
)
pipe.scheduler = UniPCMultistepScheduler.from_config(pipe.scheduler.config)
pipe.enable_model_cpu_offload()
print("ControlNet loaded.")
```

## 6. Inference
Run the model.

```
prompt = "cyberpunk version of vermeer lady, neon lights, high quality"
image = pipe(
    prompt,
    image=canny_image,
    num_inference_steps=20
).images[0]
images = [image] # For kitchen sink
print("Image generated.")
```
