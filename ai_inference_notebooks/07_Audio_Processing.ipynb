# Notebook 7: Audio Tasks

We explore AI applied to sound.

## Tasks:
1.  **Automatic Speech Recognition (ASR)**: Speech-to-Text.
2.  **Audio Classification**: Identifying sounds (e.g., "dog barking", "siren").
3.  **Text-to-Speech (TTS)**: Computer speaking.

## Setup

```python
from transformers import pipeline
from datasets import load_dataset
import soundfile as sf
import torch

# Load a sample audio for ASR
dataset = load_dataset("librispeech_asr", "clean", split="validation", streaming=True)
sample = next(iter(dataset))
audio_array = sample["audio"]["array"]
sampling_rate = sample["audio"]["sampling_rate"]
```

## Task 1: Automatic Speech Recognition (ASR)

```python
transcriber = pipeline("automatic-speech-recognition")

# We pass the raw audio
result = transcriber({"sampling_rate": sampling_rate, "raw": audio_array})
print(f"Transcription: {result['text']}")
```

## Task 2: Audio Classification

Classifying what type of sound is in the audio file.
We need a different sample (e.g., a command or environmental sound). For now, we will use a generic pipeline on the speech sample to see what it says (likely "speech").

```python
# Using a model trained on AudioSet or similar
classifier = pipeline("audio-classification", model="superb/wav2vec2-base-superb-ks")

# Note: This model expects short clips (Keyword Spotting). 
# We'll just run it on our speech sample.
result = classifier(audio_array[:16000], top_k=3) # First 1 second approx
print(result)
```

## Task 3: Text-to-Speech (TTS)

Generating audio from text. 
*Note: `text-to-speech` pipeline support in transformers is evolving.*

```python
tts = pipeline("text-to-speech", model="suno/bark-small")

text = "Hello! I am a computer generating this voice."
output = tts(text)

# The output is an audio array. We can play it in a notebook.
from IPython.display import Audio
Audio(output["audio"][0], rate=output["sampling_rate"])
```
