## Step 2: Setup and Imports

```python
from transformers import pipeline
from datasets import load_dataset
```

## Step 3: Run Inference

```python
clf = pipeline("audio-classification", model="superb/wav2vec2-base-superb-ks")

# Load sample
dataset = load_dataset("superb", "ks", split="validation", streaming=True, trust_remote_code=True)
sample = next(iter(dataset))

# The model predicts the keyword spoken
result = clf(sample["audio"]["array"])
print(f"Prediction: {result[0]['label']}, Score: {result[0]['score']:.4f}")
```
