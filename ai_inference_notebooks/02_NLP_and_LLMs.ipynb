# Notebook 2: Natural Language Processing (NLP) and LLMs

In this notebook, we will explore **Natural Language Processing (NLP)**. NLP is the field of AI focused on enabling computers to understand, interpret, and generate human language.

We will use the `pipeline` function from the `transformers` library. This is a very powerful tool that abstracts away most of the complex code, allowing us to use state-of-the-art models with just a few lines of code.

## Tasks we will cover:
1. **Sentiment Analysis**: Determining if a text is positive or negative.
2. **Text Generation**: Using a Large Language Model (LLM) to write text.
3. **Summarization**: Creating a short summary of a long text.
4. **Translation**: Translating text from one language to another.

## Setup
First, let's import the `pipeline` function.

```python
from transformers import pipeline

print("Pipeline imported successfully!")
```

## Task 1: Sentiment Analysis

Sentiment analysis is used to determine the emotional tone behind a series of words.

**How to do it:**
We create a pipeline for `"sentiment-analysis"`. The library will automatically download a default pre-trained model suitable for this task.

```python
# Create a sentiment analysis pipeline
# The first time you run this, it will download the model (approx 200-500MB).
sentiment_classifier = pipeline("sentiment-analysis")

# Let's test it!
result = sentiment_classifier("I love learning about Artificial Intelligence!")
print(f"Result 1: {result}")

result2 = sentiment_classifier("I am frustrated because the code is not working.")
print(f"Result 2: {result2}")
```

**Understanding the output:**
The output is a list containing a dictionary with:
- `label`: 'POSITIVE' or 'NEGATIVE'
- `score`: How confident the model is (between 0 and 1).

## Task 2: Text Generation (LLMs)

This is the technology behind ChatGPT and other AI assistants. We give the model a starting phrase (prompt), and it completes it.

We will use `"text-generation"`. By default, it uses GPT-2, which is an older but fast and lightweight model.

```python
# Create a text generation pipeline
generator = pipeline("text-generation", model="gpt2")

# Generate text
# max_length limits how long the generated text can be.
output = generator("Once upon a time, in a land far away,", max_length=50, num_return_sequences=1)

print(output[0]['generated_text'])
```

*Note: Since these models are trained on internet data, they can sometimes generate unpredictable or biased content.*

## Task 3: Summarization

Summarization models take a long text and produce a shorter version that keeps the main points.

We will use a pipeline for `"summarization"`.

```python
summarizer = pipeline("summarization")

long_text = """
Artificial intelligence (AI) is intelligence demonstrated by machines, as opposed to the natural intelligence displayed by animals including humans. 
AI research has been defined as the field of study of intelligent agents, which refers to any system that perceives its environment and takes actions that maximize its chance of achieving its goals.
The term "artificial intelligence" had previously been used to describe machines that mimic and display "human" cognitive skills that are associated with the human mind, such as "learning" and "problem-solving". 
This definition has since been rejected by major AI researchers who now describe AI in terms of rationality and acting rationally, which does not limit how intelligence can be articulated.
"""

summary = summarizer(long_text, max_length=60, min_length=10, do_sample=False)

print("Original length:", len(long_text))
print("Summary:")
print(summary[0]['summary_text'])
```

## Task 4: Translation

Let's translate text. We can specify the language pair, for example, English to French (`en_to_fr`) or use a specific model.

We will use a specific model `Helsinki-NLP/opus-mt-en-fr` for English to French translation to ensure it works smoothly.

```python
# English to French translation
translator = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-fr")

text_to_translate = "Hello, how are you today?"
translated = translator(text_to_translate)

print(f"Original: {text_to_translate}")
print(f"French: {translated[0]['translation_text']}")
```
