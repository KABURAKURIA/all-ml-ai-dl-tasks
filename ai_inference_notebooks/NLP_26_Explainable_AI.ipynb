## 2. Imports

```python
import shap
import torch
import transformers
import numpy as np
import matplotlib.pyplot as plt
```

## 3. Load Data

```python
data = [
    "I love this movie, it is fantastic!",
    "The plot was boring and the acting was terrible."
]
print("Data loaded.")
```

## 4. Data Cleaning

```python
# No specific cleaning needed for this demo
print("Data ready.")
```

## 5. Load Model

```python
model_id = "distilbert-base-uncased-finetuned-sst-2-english"
pipe = transformers.pipeline("text-classification", model=model_id, top_k=None)
print("Model loaded.")
```

## 6. Inference

```python
# 1. Create SHAP Explainer
explainer = shap.Explainer(pipe)

# 2. Calculate SHAP values
shap_values = explainer(data)

print("SHAP values calculated.")
# 3. Generate and Save Plot
shap.plots.bar(shap_values[0], show=False)
plt.savefig("shap_bar.png", bbox_inches='tight')
print("Saved shap_bar.png")

# Load for Kitchen Sink
try:
    from PIL import Image
    image = Image.open("shap_bar.png")
except:
    pass

results = str(shap_values)
```
