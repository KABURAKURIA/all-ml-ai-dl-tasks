## 2. Imports


```python
from transformers import pipeline
from PIL import Image
import requests
import matplotlib.pyplot as plt
import torch
```

## 3. Load Data
We will load a sample image from the COCO dataset.

```python
url = "http://images.cocodataset.org/val2017/000000039769.jpg"
image = Image.open(requests.get(url, stream=True).raw)
print("Input Image:")
display(image)
```

## 4. Data Cleaning
Standardize the image (RGB).

```python
if 'image' in locals():
    if image.mode != 'RGB':
        image = image.convert('RGB')
    print(f"Image size: {image.size}")
else:
    print("No image loaded.")
```

## 5. Load Model
Load `LiheYoung/depth-anything-small-hf`.

```python
device = 0 if torch.cuda.is_available() else -1
depth_estimator = pipeline(task="depth-estimation", model="LiheYoung/depth-anything-small-hf", device=device)
print(f"Model loaded on device: {device}")
```

## 6. Inference


```python
result = depth_estimator(image)
depth_map = result["depth"]

print("Depth Map:")
display(depth_map)
```

## 7. Save Results


```python
depth_map.save("depth_map.png")

with open('results.txt', 'w') as f:
    f.write("Depth map saved to depth_map.png")
print('Results saved.')
```
