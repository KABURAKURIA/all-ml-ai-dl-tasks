## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
from PIL import Image
import requests
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
url = "http://images.cocodataset.org/val2017/000000039769.jpg"
image = Image.open(requests.get(url, stream=True).raw)
display(image)
```

## 4. Data Cleaning
We will standardise the image by resizing it to a fixed size and ensuring it is in RGB format.
This is a common feature engineering step in Computer Vision.


```python
# Data Cleaning & Feature Engineering
if 'image' in locals():
    # Ensure RGB
    if image.mode != 'RGB':
        image = image.convert('RGB')
    # Resize to standard size (e.g., 224x224 for ResNet, though pipelines handle this, explicit resizing is good practice)
    image = image.resize((224, 224))
    print('Image resized to (224, 224) and converted to RGB.')
    display(image)
else:
    print('Variable \'image\' not found. Skipping cleaning.')
```

## 5. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
classifier = pipeline("zero-shot-image-classification", model="openai/clip-vit-base-patch32")
print("CLIP loaded.")
```

## 6. Inference
Passing our data through the model to get predictions.

```python
labels = ["cat", "dog", "couch", "remote"]
results = classifier(image, candidate_labels=labels)

for res in results:
    print(f"Label: {res['label']}, Score: {res['score']:.4f}")
```
