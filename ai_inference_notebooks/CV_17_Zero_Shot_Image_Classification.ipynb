## Step 2: Setup and Imports

```python
from transformers import pipeline
from PIL import Image
import requests
```

## Step 3: Run Inference

```python
clf = pipeline("zero-shot-image-classification", model="openai/clip-vit-base-patch32")
url = "http://images.cocodataset.org/val2017/000000039769.jpg"
image = Image.open(requests.get(url, stream=True).raw)

labels = ["cat", "dog", "car", "remote"]
results = clf(image, candidate_labels=labels)
for res in results:
    print(f"Label: {res['label']}, Score: {res['score']:.4f}")
```
