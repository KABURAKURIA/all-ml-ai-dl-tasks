## 2. Setup Agent & Model
We use `HfApiModel` which utilizes the Hugging Face Serverless Inference API. 
**Note:** You might need a Hugging Face Token (read) if you hit rate limits. Set it as `HF_TOKEN` in your environment or pass it to `HfApiModel`.

We will use `Qwen/Qwen2.5-Coder-32B-Instruct` which is excellent for code generation.

```python
from smolagents import CodeAgent, HfApiModel, tool

# You can pass 'token="YOUR_HF_TOKEN"' to HfApiModel if needed.
model = HfApiModel(model_id="Qwen/Qwen2.5-Coder-32B-Instruct")

print("Model initialized.")
```

## 3. Define Tools
We can define custom tools using the `@tool` decorator. Let's create a tool that calculates the Nth Fibonacci number efficiently.

```python
@tool
def fibonacci_tool(n: int) -> int:
    """
    Calculates the nth Fibonacci number.
    Args:
        n: The position in the Fibonacci sequence.
    """
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        a, b = 0, 1
        for _ in range(2, n + 1):
            a, b = b, a + b
        return b

print("Tool defined.")
```

## 4. Run Agent
We initialize the `CodeAgent` with our tool. It also has access to a Python interpreter by default.

```python
agent = CodeAgent(tools=[fibonacci_tool], model=model)

task = "What is the 50th Fibonacci number? Then, take that number and tell me how many digits it has."

result = agent.run(task)

print(f"\nFinal Result: {result}")
```
