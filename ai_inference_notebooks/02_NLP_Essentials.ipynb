# Notebook 2: NLP Essentials

In this notebook, we will explore core **Natural Language Processing (NLP)** tasks. These tasks involve understanding and processing human language.

We will use the `pipeline` function from the `transformers` library, which makes using these models extremely easy.

## Tasks:
1.  **Text Classification**: Sentiment analysis (Positive/Negative).
2.  **Token Classification (NER)**: Finding names, dates, and locations in text.
3.  **Question Answering**: Answering questions based on a context.
4.  **Summarization**: Summarizing long texts.
5.  **Translation**: Translating between languages.
6.  **Zero-Shot Classification**: Classifying text into custom categories without specific training.

## Setup

```python
from transformers import pipeline
import warnings
warnings.filterwarnings("ignore") # Ignore warning messages for cleaner output

print("Ready for NLP tasks!")
```

## Task 1: Text Classification (Sentiment Analysis)

We determine the emotional tone of text.

```python
classifier = pipeline("text-classification")

result = classifier("I am absolutely delighted with these results!")
print(f"Result: {result}")
```

## Task 2: Token Classification (Named Entity Recognition)

We identify "entities" like people (PER), organizations (ORG), and locations (LOC).

```python
ner = pipeline("token-classification", aggregation_strategy="simple")

text = "Apple was founded by Steve Jobs in California."
entities = ner(text)

for entity in entities:
    print(f"Entity: {entity['word']}, Type: {entity['entity_group']}, Score: {entity['score']:.2f}")
```

## Task 3: Question Answering

We provide a `context` (a paragraph of information) and a `question`. The model finds the answer in the context.

```python
qa_model = pipeline("question-answering")

context = "The capital of France is Paris. It is known for the Eiffel Tower."
question = "What is the capital of France?"

answer = qa_model(question=question, context=context)
print(f"Answer: {answer['answer']}")
```

## Task 4: Summarization

We condense a long text into a shorter one.

```python
summarizer = pipeline("summarization")

long_text = """
Artificial intelligence (AI) is intelligence demonstrated by machines, as opposed to the natural intelligence displayed by animals including humans. 
AI research has been defined as the field of study of intelligent agents, which refers to any system that perceives its environment and takes actions that maximize its chance of achieving its goals.
The term "artificial intelligence" had previously been used to describe machines that mimic and display "human" cognitive skills that are associated with the human mind, such as "learning" and "problem-solving".
"""

summary = summarizer(long_text, max_length=50, min_length=10)
print(f"Summary: {summary[0]['summary_text']}")
```

## Task 5: Translation

Translating text. We specify the model for English to French.

```python
translator = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-fr")

text = "Hello, how are you today?"
translated = translator(text)
print(f"Translated: {translated[0]['translation_text']}")
```

## Task 6: Zero-Shot Classification

This is magic! You can classify text into ANY categories you define, without the model having seen them before.

```python
classifier = pipeline("zero-shot-classification")

text = "The galaxy contains billions of stars and planets."
candidate_labels = ["politics", "science", "sports"]

result = classifier(text, candidate_labels)
print(f"Text: {text}")
print(f"Top Label: {result['labels'][0]}")
print(f"Scores: {result['scores']}")
```
