## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
from PIL import Image
import requests
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
# Download a sample document image (e.g., a scientific paper page or invoice)
url = "https://huggingface.co/datasets/hf-internal-testing/fixtures_docvqa/resolve/main/invoice.png"
image = Image.open(requests.get(url, stream=True).raw)

display(image.resize((300, 400)))
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
# Load LayoutLMv3 or DiT (Document Image Transformer) for classification
# We use a DiT model fine-tuned on RVL-CDIP
classifier = pipeline("image-classification", model="microsoft/dit-base-finetuned-rvl-cdip")
print("DiT model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
results = classifier(image)

print("Classification Results:")
for res in results:
    print(f"Label: {res['label']}, Confidence: {res['score']:.4f}")
```
