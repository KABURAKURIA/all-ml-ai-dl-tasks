## 2. Imports
Importing the necessary Python modules.

```python
from sentence_transformers import CrossEncoder
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
query = "What is the best way to learn Python?"
documents = [
    "Python is a popular programming language known for its simplicity.",
    "The python snake is a constrictor found in tropics.",
    "Learning to code requires practice and patience.",
    "To learn Python, start with tutorials and build small projects."
]
print(f"Query: {query}")
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
model = CrossEncoder('cross-encoder/ms-marco-TinyBERT-L-2')
print("CrossEncoder loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
# Predict scores for (query, doc) pairs
pairs = [[query, doc] for doc in documents]
scores = model.predict(pairs)

# Sort
ranked = sorted(zip(scores, documents), key=lambda x: x[0], reverse=True)

print("Ranked Documents:")
for score, doc in ranked:
    print(f"Score: {score:.4f} | Doc: {doc}")
```
