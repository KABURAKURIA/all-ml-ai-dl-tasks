## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
# Load image
url = "http://images.cocodataset.org/val2017/000000039769.jpg"
image = Image.open(requests.get(url, stream=True).raw)
display(image)
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
# Load VQA pipeline (ViLT - Vision-and-Language Transformer)
vqa = pipeline("visual-question-answering", model="dandelin/vilt-b32-finetuned-vqa")
print("ViLT model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
questions = [
    "How many cats are there?",
    "What are the cats doing?",
    "Is there a dog?"
]

for q in questions:
    res = vqa(image, q, top_k=1)
    print(f"Q: {q} | A: {res[0]['answer']} (Conf: {res[0]['score']:.2f})")
```
