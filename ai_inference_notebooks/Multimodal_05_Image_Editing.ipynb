## Step 2: Setup and Imports

```python
from diffusers import StableDiffusionInstructPix2PixPipeline
import torch
from PIL import Image
import requests
from io import BytesIO
```

## Step 3: Run Inference

```python
# Load the pipeline
device = "cuda" if torch.cuda.is_available() else "cpu"
model_id = "timbrooks/instruct-pix2pix"

# Use float16 if on GPU to save memory
dtype = torch.float16 if device == "cuda" else torch.float32
pipe = StableDiffusionInstructPix2PixPipeline.from_pretrained(model_id, torch_dtype=dtype).to(device)

# Download an image
url = "https://raw.githubusercontent.com/timothybrooks/instruct-pix2pix/main/imgs/example.jpg"
response = requests.get(url)
image = Image.open(BytesIO(response.content)).convert("RGB").resize((512, 512))

prompt = "turn him into a cyborg"
result = pipe(prompt=prompt, image=image, num_inference_steps=10 if device=="cpu" else 20).images[0]

print("Displaying result (save to see file):")
result.save("edited_image.png")
result
```
