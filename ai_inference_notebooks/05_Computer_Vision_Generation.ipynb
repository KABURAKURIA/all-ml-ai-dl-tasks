## Setup

```python
from diffusers import StableDiffusionPipeline
import torch

# Check if GPU is available to make it faster
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

## Task 1: Text-to-Image

We use **Stable Diffusion**.
*Warning: This downloads a large model (several GBs) the first time you run it.*

```python
# We use a small version/variant for demonstration if possible, 
# or standard Stable Diffusion v1.4/1.5/2.1
model_id = "CompVis/stable-diffusion-v1-4"

# If you have low RAM, you might need to adjust settings or use 'float16' revision
try:
    pipe = StableDiffusionPipeline.from_pretrained(model_id)
    if device == "cuda":
        pipe = pipe.to(device)
    
    prompt = "a photograph of an astronaut riding a horse on mars"
    image = pipe(prompt).images[0]
    
    # Display
    image.save("astronaut_mars.png")
    image
except Exception as e:
    print(f"Could not run generation: {e}")
    print("This might be due to memory constraints on your machine.")
```
