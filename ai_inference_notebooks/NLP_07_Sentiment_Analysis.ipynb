## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
from datasets import load_dataset
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
# Load samples from IMDB dataset
dataset = load_dataset("imdb", split="test", streaming=True)
samples = [next(iter(dataset)) for _ in range(5)]
texts = [s['text'][:512] for s in samples] # Truncate for efficiency

print(f"Loaded {len(texts)} samples.")
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
print("Model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
results = classifier(texts)

for text, res in zip(texts, results):
    print(f"Label: {res['label']}, Score: {res['score']:.4f}")
    print(f"Snippet: {text[:100]}...")
    print("-" * 30)
```
