# Notebook 3: NLP Advanced & LLMs

In this notebook, we look at more advanced NLP tasks, including Large Language Models (LLMs).

## Tasks:
1.  **Text Generation (LLMs)**: Creating text (like GPT).
2.  **Fill-Mask**: Guessing missing words in a sentence.
3.  **Table Question Answering**: Answering questions based on tabular data.
4.  **Sentence Similarity**: Finding how similar two sentences are.
5.  **Feature Extraction**: Converting text into numbers (vectors) for machine learning.

## Setup

```python
from transformers import pipeline
from sentence_transformers import SentenceTransformer, util
import warnings
warnings.filterwarnings("ignore")

print("Ready for Advanced NLP!")
```

## Task 1: Text Generation (LLMs)

We use a generative model to complete a prompt. We'll use `gpt2` (small and fast) for demonstration.

```python
generator = pipeline("text-generation", model="gpt2")

prompt = "The future of AI is"
output = generator(prompt, max_length=50, num_return_sequences=1)

print(output[0]['generated_text'])
```

## Task 2: Fill-Mask

The model guesses the masked word (often denoted by `<mask>`). We use BERT-based models here.

```python
unmasker = pipeline("fill-mask", model="bert-base-uncased")

# BERT uses [MASK] token
text = "The capital of France is [MASK]."
results = unmasker(text)

for res in results:
    print(f"Word: {res['token_str']}, Score: {res['score']:.4f}")
```

## Task 3: Table Question Answering

Answering questions where the context is a table (like a spreadsheet row).

```python
import pandas as pd

table_qa = pipeline("table-question-answering", model="google/tapas-base-finetuned-wtq")

data = {
    "Actors": ["Brad Pitt", "Leonardo Di Caprio", "George Clooney"],
    "Age": ["57", "46", "60"],
    "Number of movies": ["87", "53", "69"]
}

# Questions
query = "How old is Leonardo Di Caprio?"

result = table_qa(table=data, query=query)
print(f"Answer: {result['answer']}")
```

## Task 4: Sentence Similarity

We calculate a score indicating how semantically similar two sentences are using `sentence-transformers`.

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

sentences = ["The cat sits outside", "A man is playing guitar", "The feline is outdoors"]
embeddings = model.encode(sentences)

# Compute cosine similarity
# Compare sentence 0 with sentence 2
score = util.cos_sim(embeddings[0], embeddings[2])
print(f"Similarity between '{sentences[0]}' and '{sentences[2]}': {score.item():.4f}")

# Compare sentence 0 with sentence 1
score_diff = util.cos_sim(embeddings[0], embeddings[1])
print(f"Similarity between '{sentences[0]}' and '{sentences[1]}': {score_diff.item():.4f}")
```

## Task 5: Feature Extraction

Converting text into vectors (lists of numbers). This is useful if you want to feed text into another machine learning model.

```python
extractor = pipeline("feature-extraction", model="distilbert-base-cased")

text = "Hello world"
features = extractor(text)

# The output is a multi-dimensional array representing the text
print(f"Feature vector shape: {len(features[0])} tokens x {len(features[0][0])} dimensions")
print("First 5 numbers of the first token:", features[0][0][:5])
```
