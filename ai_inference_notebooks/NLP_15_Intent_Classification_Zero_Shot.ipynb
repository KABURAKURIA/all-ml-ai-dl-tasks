## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
query = "My internet is not working, can you help?"
labels = ["billing_issue", "technical_support", "sales_inquiry", "password_reset"]
print(f"Query: {query}")
print(f"Possible Intents: {labels}")
```

## Data Cleaning & Feature Engineering
We clean the text input by removing excessive whitespace and normalizing case.
This ensures the model receives consistent input.


```python
import re

def clean_text(text):
    # Remove extra spaces and strip
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Apply to our input variable (assuming 'prompt', 'text', or 'text_samples' exists)
if 'prompt' in locals():
    prompt = clean_text(prompt)
    print(f'Cleaned Prompt: {prompt}')
elif 'text' in locals():
    text = clean_text(text)
    print(f'Cleaned Text: {text}')
elif 'text_samples' in locals() and isinstance(text_samples, list):
    text_samples = [clean_text(t) for t in text_samples]
    print('Cleaned Text Samples.')
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
print("Model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
result = classifier(query, labels)
print(f"Predicted Intent: {result['labels'][0]} (Score: {result['scores'][0]:.4f})")
```
