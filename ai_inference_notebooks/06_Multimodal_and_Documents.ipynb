# Notebook 6: Multimodal & Documents

Multimodal tasks involve more than one type of data (e.g., Image + Text).

## Tasks:
1.  **Image Captioning (Image-to-Text)**: Writing a description of an image.
2.  **Visual Question Answering (VQA)**: Asking questions about an image.
3.  **Document Question Answering**: Reading a scanned document and answering questions about it.

## Setup

```python
from transformers import pipeline
from PIL import Image
import requests

def get_image(url):
    return Image.open(requests.get(url, stream=True).raw)

image_url = "http://images.cocodataset.org/val2017/000000039769.jpg"
image = get_image(image_url)
image
```

## Task 1: Image Captioning

The model looks at the image and writes a sentence.

```python
captioner = pipeline("image-to-text")

result = captioner(image)
print(result[0]['generated_text'])
```

## Task 2: Visual Question Answering (VQA)

We ask: 'How many cats are there?'

```python
vqa = pipeline("visual-question-answering")

question = "How many cats are there?"
result = vqa(image, question, top_k=1)

print(f"Question: {question}")
print(f"Answer: {result[0]['answer']}")
```

## Task 3: Document Question Answering

This is useful for receipts, invoices, or scanned PDF pages. We use a model designed to understand document layout (like LayoutLM or Donut).

```python
doc_qa = pipeline("document-question-answering", model="naver-clova-ix/donut-base-finetuned-docvqa")

# We need a document image. Let's get a sample receipt.
doc_url = "https://templates.invoicehome.com/invoice-template-us-neat-750px.png"
doc_image = get_image(doc_url)

question = "What is the total amount?"
result = doc_qa(image=doc_image, question=question)

print(f"Answer: {result[0]['answer']}")
doc_image
```
