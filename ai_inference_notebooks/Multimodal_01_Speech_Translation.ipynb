## Step 2: Setup and Imports

```python
from transformers import pipeline
from datasets import load_dataset
import torch
```

## Step 3: Run Inference

```python
# We use OpenAI's Whisper model which is excellent for speech translation
# It translates various languages to English
device = "cuda" if torch.cuda.is_available() else "cpu"
translator = pipeline("automatic-speech-recognition", model="openai/whisper-tiny", device=device)

# Load a sample French audio
dataset = load_dataset("common_voice", "fr", split="validation", streaming=True, trust_remote_code=True)
sample = next(iter(dataset))
audio = sample["audio"]

# Run inference with task="translate"
result = translator(audio["array"], return_timestamps=True, generate_kwargs={"task": "translate"})

print("Original Audio (French) translated to English:")
print(result["text"])
```
