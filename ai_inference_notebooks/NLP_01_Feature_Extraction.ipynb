## Data Cleaning & Feature Engineering
We clean the text input by removing excessive whitespace and normalizing case.
This ensures the model receives consistent input.


```python
import re

def clean_text(text):
    # Remove extra spaces and strip
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Apply to our input variable (assuming 'prompt', 'text', or 'text_samples' exists)
if 'prompt' in locals():
    prompt = clean_text(prompt)
    print(f'Cleaned Prompt: {prompt}')
elif 'text' in locals():
    text = clean_text(text)
    print(f'Cleaned Text: {text}')
elif 'text_samples' in locals() and isinstance(text_samples, list):
    text_samples = [clean_text(t) for t in text_samples]
    print('Cleaned Text Samples.')
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
# Load a feature extraction pipeline (DistilBERT)
extractor = pipeline("feature-extraction", model="distilbert-base-cased")
print("Model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
# Extract features
# The output is a list of lists (batch x sequence_length x hidden_size)
features = extractor(text_samples)

print(f"Number of samples processed: {len(features)}")
print(f"Embedding shape for first sample: {len(features[0])} tokens x {len(features[0][0])} dimensions")
print("First 5 values of the CLS token (first token) of the first sentence:")
print(features[0][0][:5])
```
