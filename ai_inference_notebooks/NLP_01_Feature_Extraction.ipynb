## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline, AutoTokenizer, AutoModel
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
text_samples = [
    "Artificial Intelligence is fascinating.",
    "The weather is sunny today.",
    "Deep learning transforms industries."
]
print("Input Texts:", text_samples)
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
# Load a feature extraction pipeline (DistilBERT)
extractor = pipeline("feature-extraction", model="distilbert-base-cased")
print("Model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
# Extract features
# The output is a list of lists (batch x sequence_length x hidden_size)
features = extractor(text_samples)

print(f"Number of samples processed: {len(features)}")
print(f"Embedding shape for first sample: {len(features[0])} tokens x {len(features[0][0])} dimensions")
print("First 5 values of the CLS token (first token) of the first sentence:")
print(features[0][0][:5])
```
