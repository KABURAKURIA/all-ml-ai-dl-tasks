## 2. Imports

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
# Generate synthetic data
# y = 2x + 1 + noise
X = np.random.rand(100, 1) * 10
y = 2 * X + 1 + np.random.randn(100, 1) * 0.5

df = pd.DataFrame({'X': X.flatten(), 'y': y.flatten()})
print("First 5 rows:")
display(df.head())
```

## Data Cleaning & Feature Engineering
We impute missing values and scale numerical features.


```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

if 'X' in locals() and isinstance(X, pd.DataFrame):
    # 1. Cleaning
    imputer = SimpleImputer(strategy='mean')
    X_clean = pd.DataFrame(imputer.fit_transform(X), columns=X.columns)
    
    # 2. Feature Engineering
    scaler = StandardScaler()
    X_scaled = pd.DataFrame(scaler.fit_transform(X_clean), columns=X.columns)
    
    X = X_scaled
    print('Data imputed and scaled.')
    display(X.head())
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
model = LinearRegression()
print("LinearRegression initialized.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
model.fit(X, y)
print(f"Learned Coefficient: {model.coef_[0][0]:.2f} (Expected ~2.0)")
print(f"Learned Intercept: {model.intercept_[0]:.2f} (Expected ~1.0)")

test_val = np.array([[5.0]])
pred = model.predict(test_val)
print(f"Prediction for X=5.0: {pred[0][0]:.2f}")
```
