## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
from huggingface_hub import hf_hub_download
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
# Get a video file
video_path = hf_hub_download(repo_id="nateraw/video-vit-age-v2", filename="archery.mp4")
print(f"Video path: {video_path}")
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
classifier = pipeline("video-classification", model="nateraw/video-vit-age-v2")
print("Model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
result = classifier(video_path)
print("Predictions:")
for r in result:
    print(f"{r['label']}: {r['score']:.4f}")
```
