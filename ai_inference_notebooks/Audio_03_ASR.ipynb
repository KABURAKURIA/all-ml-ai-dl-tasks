## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
from datasets import load_dataset
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
# Load LibriSpeech sample
dataset = load_dataset("patrickvonplaten/librispeech_asr_dummy", "clean", split="validation", trust_remote_code=True)
sample = next(iter(dataset))
audio = sample["audio"]["array"]
print("Audio loaded.")
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
transcriber = pipeline("automatic-speech-recognition", model="facebook/wav2vec2-base-960h")
print("Wav2Vec2 model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
result = transcriber(audio)
print(f"Transcription: {result['text']}")
```
