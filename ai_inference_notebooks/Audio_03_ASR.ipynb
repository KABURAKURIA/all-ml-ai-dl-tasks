## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
from datasets import load_dataset
import torch
import librosa
import numpy as np
```

## 3. Load Data
We will load a dummy sample from the LibriSpeech dataset.

```python
dataset = load_dataset("patrickvonplaten/librispeech_asr_dummy", "clean", split="validation", trust_remote_code=True)
sample = next(iter(dataset))
audio = sample["audio"]["array"]
sampling_rate = sample["audio"]["sampling_rate"]
print(f"Audio loaded. Sampling Rate: {sampling_rate}Hz")
```

## Data Cleaning & Feature Engineering
Although the Whisper pipeline handles resampling, it is best practice to ensure our input matches the standard 16kHz used in speech tasks.


```python
target_sr = 16000
if sampling_rate != target_sr:
    audio = librosa.resample(audio, orig_sr=sampling_rate, target_sr=target_sr)
    sampling_rate = target_sr
    print(f'Resampled audio to {target_sr}Hz.')
else:
    print(f'Audio is already at {target_sr}Hz.')
```

## 4. Load Model
We will load **openai/whisper-small**. Whisper is a general-purpose speech recognition model. It achieves state-of-the-art performance on many benchmarks.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
transcriber = pipeline("automatic-speech-recognition", model="openai/whisper-small", device=device)
print("Whisper Small loaded.")
```

## 5. Inference
Pass the processed audio to the model.

```python
# The pipeline expects a dictionary for raw audio
input_audio = {"raw": audio, "sampling_rate": sampling_rate}
result = transcriber(input_audio)
print(f"Transcription: {result['text']}")
```
