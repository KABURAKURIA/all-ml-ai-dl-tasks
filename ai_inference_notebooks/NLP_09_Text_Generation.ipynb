## 2. Imports
Importing the necessary Python modules.

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
```

## 3. Load Model
We load **HuggingFaceTB/SmolLM2-1.7B-Instruct**. This is a state-of-the-art small language model that fits easily on a T4 GPU in fp16 precision.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
model_id = "HuggingFaceTB/SmolLM2-1.7B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch.float16).to(device)

print("SmolLM2 loaded.")
```

## 4. Prepare Input
Since this is an instruction-tuned model, we use a chat template to format our prompt.

```python
messages = [{"role": "user", "content": "Write a short story about a robot learning to paint."}]
input_text = tokenizer.apply_chat_template(messages, tokenize=False)
print(f"Input Text:\n{input_text}")
```

## 5. Inference
Generate the response.

```python
inputs = tokenizer(input_text, return_tensors="pt").to(device)
outputs = model.generate(**inputs, max_new_tokens=200, temperature=0.7, top_p=0.9, do_sample=True)
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(f"Generated Text:\n{generated_text}")
```
