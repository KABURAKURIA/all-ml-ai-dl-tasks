Classify text into topics using Zero-Shot Classification.

## Step 2: Setup and Imports

```python
from transformers import pipeline
```

## Step 3: Run Inference

```python
classifier = pipeline("zero-shot-classification")
text = "The election results were announced yesterday."
labels = ["politics", "sports", "technology", "entertainment"]

result = classifier(text, labels)
print(f"Text: {text}")
print(f"Top Topic: {result['labels'][0]}")
```
