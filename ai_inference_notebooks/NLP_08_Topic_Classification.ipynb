## 3. Load Data

```python
text = "The stock market crashed today due to new inflation data."
candidate_labels = ["politics", "economy", "sports", "technology"]
print(f"Text: {text}")
print(f"Candidates: {candidate_labels}")
```

## Data Cleaning & Feature Engineering
We clean the text input by removing excessive whitespace and normalizing case.
This ensures the model receives consistent input.


```python
import re

def clean_text(text):
    # Remove extra spaces and strip
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Apply to our input variable
if 'prompt' in locals():
    prompt = clean_text(prompt)
    print(f'Cleaned Prompt: {prompt}')
elif 'text' in locals():
    text = clean_text(text)
    print(f'Cleaned Text: {text}')
elif 'texts' in locals() and isinstance(texts, list):
    texts = [clean_text(t) for t in texts]
    print(f'Cleaned {len(texts)} text samples.')
elif 'text_samples' in locals() and isinstance(text_samples, list):
    text_samples = [clean_text(t) for t in text_samples]
    print(f'Cleaned {len(text_samples)} text samples.')
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
print("BART MNLI model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
result = classifier(text, candidate_labels)

print("Scores:")
for label, score in zip(result['labels'], result['scores']):
    print(f"{label}: {score:.4f}")
```
