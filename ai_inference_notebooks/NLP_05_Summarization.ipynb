## Step 2: Setup and Imports

```python
from transformers import pipeline
```

## Step 3: Run Inference

```python
summarizer = pipeline("summarization")
text = """
The Transformers library is an open-source library for natural language processing (NLP) tasks. 
It provides state-of-the-art models for various tasks such as text classification, information extraction, question answering, summarization, translation, and text generation. 
The library is built on top of PyTorch and TensorFlow, making it easy to use and integrate into existing workflows. 
It is maintained by Hugging Face and has a large community of contributors.
"""
result = summarizer(text, max_length=50, min_length=10)
print(f"Summary: {result[0]['summary_text']}")
```
