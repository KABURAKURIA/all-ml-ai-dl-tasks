## 2. Imports

```python
from transformers import pipeline
from datasets import load_dataset
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
# Load a sample from CNN/DailyMail dataset
dataset = load_dataset("cnn_dailymail", "3.0.0", split="validation", streaming=True)
sample = next(iter(dataset))
article = sample["article"]

print(f"Original Article Length: {len(article)} chars")
print("First 500 chars:", article[:500] + "...")
```

## 4. Data Cleaning
We clean the text input by removing excessive whitespace and normalizing case.
This ensures the model receives consistent input.


```python
import re

def clean_text(text):
    # Remove extra spaces and strip
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Apply to our input variable
if 'prompt' in locals():
    prompt = clean_text(prompt)
    print(f'Cleaned Prompt: {prompt}')
elif 'text' in locals():
    text = clean_text(text)
    print(f'Cleaned Text: {text}')
elif 'texts' in locals() and isinstance(texts, list):
    texts = [clean_text(t) for t in texts]
    print(f'Cleaned {len(texts)} text samples.')
elif 'text_samples' in locals() and isinstance(text_samples, list):
    text_samples = [clean_text(t) for t in text_samples]
    print(f'Cleaned {len(text_samples)} text samples.')
```

## 5. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
print("BART model loaded.")
```

## 6. Inference
Passing our data through the model to get predictions.

```python
# Summarize
# We constrain the summary length
summary = summarizer(article, max_length=130, min_length=30, do_sample=False)

print("
Generated Summary:")
print(summary[0]['summary_text'])
```
