## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
from datasets import load_dataset
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
# Load a sample from CNN/DailyMail dataset
dataset = load_dataset("cnn_dailymail", "3.0.0", split="validation", streaming=True)
sample = next(iter(dataset))
article = sample["article"]

print(f"Original Article Length: {len(article)} chars")
print("First 500 chars:", article[:500] + "...")
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
print("BART model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
# Summarize
# We constrain the summary length
summary = summarizer(article, max_length=130, min_length=30, do_sample=False)

print("
Generated Summary:")
print(summary[0]['summary_text'])
```
