## Step 3: Run Inference

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
emb1 = model.encode("This is a happy dog")
emb2 = model.encode("This is a joyous canine")
print(util.cos_sim(emb1, emb2))
```
