## 2. Imports
Importing the necessary Python modules.

```python
from sentence_transformers import SentenceTransformer, util
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
sentences = [
    "The cat sits outside",
    "A man is playing guitar",
    "The feline is outdoors",
    "A person performs music"
]
print("Sentences:", sentences)
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
print("Sentence Transformer loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
embeddings = model.encode(sentences)

# Compute cosine similarity
# 1. Cat vs Feline
sim1 = util.cos_sim(embeddings[0], embeddings[2])
print(f"Similarity ('{sentences[0]}' vs '{sentences[2]}'): {sim1.item():.4f}")

# 2. Guitar vs Music
sim2 = util.cos_sim(embeddings[1], embeddings[3])
print(f"Similarity ('{sentences[1]}' vs '{sentences[3]}'): {sim2.item():.4f}")

# 3. Cat vs Guitar
sim3 = util.cos_sim(embeddings[0], embeddings[1])
print(f"Similarity ('{sentences[0]}' vs '{sentences[1]}'): {sim3.item():.4f}")
```
