## Step 2: Setup and Imports

```python
from sentence_transformers import SentenceTransformer, util
```

## Step 3: Run Inference

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

sentences = ["The cat sits outside", "A man is playing guitar", "The feline is outdoors"]
embeddings = model.encode(sentences)

# Compute cosine similarity between first and third sentence
score = util.cos_sim(embeddings[0], embeddings[2])
print(f"Similarity between '{sentences[0]}' and '{sentences[2]}': {score.item():.4f}")
```
