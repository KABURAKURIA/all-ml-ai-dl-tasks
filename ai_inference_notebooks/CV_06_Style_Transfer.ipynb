## Step 2: Setup and Imports

```python
from diffusers import StableDiffusionInstructPix2PixPipeline
import torch
from PIL import Image
import requests
from io import BytesIO
```

## Step 3: Run Inference

```python
# Modern style transfer is often done via instruction editing
device = "cuda" if torch.cuda.is_available() else "cpu"
pipe = StableDiffusionInstructPix2PixPipeline.from_pretrained("timbrooks/instruct-pix2pix", torch_dtype=torch.float16 if device=="cuda" else torch.float32).to(device)

url = "https://raw.githubusercontent.com/timothybrooks/instruct-pix2pix/main/imgs/example.jpg"
image = Image.open(requests.get(url, stream=True).raw).convert("RGB").resize((512, 512))

prompt = "make it look like a van gogh painting"
result = pipe(prompt=prompt, image=image).images[0]
result
```
