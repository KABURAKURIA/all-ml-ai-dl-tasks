## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
text = "Apple Inc. is planning to open a new store in San Francisco near the Golden Gate Bridge."
print(f"Input: {text}")
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
ner = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english", aggregation_strategy="simple")
print("BERT NER model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
entities = ner(text)

print("Detected Entities:")
for ent in entities:
    print(f"Entity: {ent['word']}, Type: {ent['entity_group']}, Score: {ent['score']:.4f}")
```
