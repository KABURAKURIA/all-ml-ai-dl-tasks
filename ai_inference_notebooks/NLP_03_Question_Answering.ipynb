## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
context = """
The Amazon rainforest, alternatively, the Amazon Jungle or Amazonia, is a moist broadleaf tropical rainforest in the Amazon biome that covers most of the Amazon basin of South America. This basin encompasses 7,000,000 km2 (2,700,000 sq mi), of which 5,500,000 km2 (2,100,000 sq mi) are covered by the rainforest. This region includes territory belonging to nine nations.
"""
question = "How huge is the Amazon rainforest?"
print(f"Context length: {len(context)} chars")
print(f"Question: {question}")
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
qa_model = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
print("QA Model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
result = qa_model(question=question, context=context)
print(f"Answer: {result['answer']}")
print(f"Score: {result['score']:.4f}")
print(f"Start/End Indices: {result['start']}, {result['end']}")
```
