## 2. Imports
Importing the necessary Python modules.

```python
from transformers import pipeline
import pandas as pd
```

## 3. Load Data
We will load a sample from a real dataset or prepare a complex input example to test the model.

```python
data = {
    "City": ["New York", "Los Angeles", "Chicago", "Houston"],
    "Population": ["8,336,817", "3,979,576", "2,693,976", "2,320,268"],
    "Area (sq mi)": ["302.6", "468.7", "227.3", "637.4"]
}
table = pd.DataFrame.from_dict(data)
print("Table Data:")
display(table)
```

## 4. Load Model
We will load a pre-trained model from the Hugging Face Hub.

```python
tqa = pipeline("table-question-answering", model="google/tapas-base-finetuned-wtq")
print("TAPAS model loaded.")
```

## 5. Inference
Passing our data through the model to get predictions.

```python
queries = [
    "Which city has the largest population?",
    "What is the area of Houston?",
    "List all cities."
]

for q in queries:
    res = tqa(table=table, query=q)
    print(f"Q: {q} | A: {res['answer']}")
```
