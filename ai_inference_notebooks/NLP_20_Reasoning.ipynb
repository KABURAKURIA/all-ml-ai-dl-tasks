## 2. Imports
Import modules.

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
```

## 3. Load Data
Define a logic puzzle.

```python
problem = "If a doctor gives you 3 pills and tells you to take one every half hour, how long would it be before all the pills had been taken?"
```

## 4. Data Cleaning
Prepare the prompt.

```python
# Standard Prompt
prompt_std = f"Question: {problem}\nAnswer:"

# CoT Prompt
prompt_cot = f"Question: {problem}\nLet's think step by step.\n"
```

## 5. Load Model
Load `Qwen/Qwen2.5-1.5B-Instruct` (Strong reasoning capabilities).

```python
model_id = "Qwen/Qwen2.5-1.5B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch.float16, device_map="cuda")
print("Model loaded.")
```

## 6. Inference
Compare Standard vs CoT.

```python
# Run CoT
messages = [
    {"role": "system", "content": "You are a helpful assistant. Solve the puzzle step by step."},
    {"role": "user", "content": problem}
]
text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
inputs = tokenizer(text, return_tensors="pt").to(model.device)

outputs = model.generate(**inputs, max_new_tokens=200)
response = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(f"Response:\n{response}")
```
