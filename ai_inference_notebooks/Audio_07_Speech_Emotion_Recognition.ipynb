## 5. Load Model
Load the model.

```
classifier = foreign_class(source="speechbrain/emotion-recognition-wav2vec2-IEMOCAP", pymodule_file="custom_interface.py", run_opts={"device":"cuda" if torch.cuda.is_available() else "cpu"})
print("Classifier loaded.")
```

## 6. Inference
Run the model.

```
out_prob, score, index, text_lab = classifier.classify_file("sample.wav")
print(f"Emotion: {text_lab}")
results = text_lab
```
