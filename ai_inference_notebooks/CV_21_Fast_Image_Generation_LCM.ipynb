## 2. Imports
Import DiffusionPipeline.

```python
from diffusers import DiffusionPipeline
import torch
import matplotlib.pyplot as plt
```

## 3. Load Data
We define the prompt.

```python
prompt = "A futuristic city on Mars, cinematic lighting, 8k resolution"
```

## 4. Data Cleaning

## 5. Load Model
Load `SimianLuo/LCM_Dreamshaper_v7`.

```python
pipe = DiffusionPipeline.from_pretrained("SimianLuo/LCM_Dreamshaper_v7")
pipe.to(torch_device="cuda", torch_dtype=torch.float16)
print("LCM Model loaded.")
```

## 6. Inference
Generate image with only 4 steps.

```python
image = pipe(prompt=prompt, num_inference_steps=4, guidance_scale=8.0).images[0]
plt.imshow(image)
plt.axis('off')
plt.show()
```
