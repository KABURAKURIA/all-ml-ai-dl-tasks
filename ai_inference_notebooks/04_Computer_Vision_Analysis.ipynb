# Notebook 4: Computer Vision - Understanding

In this notebook, we focus on tasks where the AI "sees" and analyzes an image.

## Tasks:
1.  **Image Classification**: What is in the picture?
2.  **Object Detection**: Where are the objects?
3.  **Image Segmentation**: Pixel-level understanding.
4.  **Zero-Shot Image Classification**: Classifying images with custom labels.
5.  **Depth Estimation**: How far away are things?

## Setup

```python
from transformers import pipeline
from PIL import Image
import requests

# Helper to download images
def get_image(url):
    return Image.open(requests.get(url, stream=True).raw)

# Cat image
image_url = "http://images.cocodataset.org/val2017/000000039769.jpg"
image = get_image(image_url)
image
```

## Task 1: Image Classification

```python
classifier = pipeline("image-classification")
results = classifier(image)

for res in results:
    print(f"Label: {res['label']}, Score: {res['score']:.4f}")
```

## Task 2: Object Detection

```python
detector = pipeline("object-detection")
detections = detector(image)

for det in detections:
    print(f"Object: {det['label']}, Box: {det['box']}")
```

## Task 3: Image Segmentation

Segmentation creates a "mask" for objects. We'll use a `mask-generation` or segmentation pipeline.

```python
segmenter = pipeline("image-segmentation", model="facebook/detr-resnet-50-panoptic")
results = segmenter(image)

print(f"Found {len(results)} segments.")
# Each result contains a 'mask' (Pillow image) and label
results[0]['mask']
```

## Task 4: Zero-Shot Image Classification

Like zero-shot text classification, we provide our own labels (e.g., "cats", "dogs", "aliens"). The model (CLIP) decides which one fits best.

```python
# We use a CLIP model
zero_shot = pipeline("zero-shot-image-classification", model="openai/clip-vit-base-patch32")

labels = ["cat", "dog", "truck"]
results = zero_shot(image, candidate_labels=labels)

for res in results:
    print(f"Label: {res['label']}, Score: {res['score']:.4f}")
```

## Task 5: Depth Estimation

The model estimates the distance of each pixel from the camera. Lighter pixels are usually closer, darker are further.

```python
depth_estimator = pipeline("depth-estimation")
result = depth_estimator(image)

# The result contains a 'depth' image
result['depth']
```
