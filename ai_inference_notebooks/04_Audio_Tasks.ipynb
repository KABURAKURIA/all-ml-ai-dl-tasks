## Task 1: Automatic Speech Recognition (ASR)

This task involves transcribing a recording of speech into text. We will use the `"automatic-speech-recognition"` pipeline.
We will use a small version of the Facebook Wav2Vec2 model, or OpenAI's Whisper model if available. The pipeline defaults are usually good.

```python
from datasets import load_dataset
import soundfile as sf

# Load a sample dataset to get an audio file
# This downloads a tiny dataset of English speech
dataset = load_dataset("librispeech_asr", "clean", split="validation", streaming=True)
sample = next(iter(dataset))

# The 'audio' field contains the array and sampling rate
audio_array = sample["audio"]["array"]
sampling_rate = sample["audio"]["sampling_rate"]

print(f"Loaded audio sample. Sampling rate: {sampling_rate} Hz")
```

Now let's run the speech recognition model on this audio.

```python
# Create the pipeline
transcriber = pipeline("automatic-speech-recognition")

# Transcribe the audio
# The pipeline handles the raw audio array directly
result = transcriber({"sampling_rate": sampling_rate, "raw": audio_array})

print("Transcription:")
print(result["text"])
```

## Comparison with Ground Truth
Let's see what the actual text was (provided in the dataset) to check how good our model is.

```python
print("Ground Truth:")
print(sample["text"])
```
