## Step 2: Setup and Imports

```python
from transformers import pipeline
from PIL import Image
import requests
```

## Step 3: Run Inference

```python
# Using Mask2Former or similar via image-segmentation pipeline which supports panoptic/instance
# We'll use a MaskFormer model fine-tuned for instance segmentation on COCO
seg = pipeline("image-segmentation", model="facebook/maskformer-swin-base-coco")
url = "http://images.cocodataset.org/val2017/000000039769.jpg"
image = Image.open(requests.get(url, stream=True).raw)

results = seg(image)
print(f"Found {len(results)} segments.")
# Display first mask
results[0]['mask']
```
