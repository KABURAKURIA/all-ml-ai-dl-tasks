## 5. Load Model
We use `cvssp/audioldm-s-full-v2` which is lightweight and capable.

```
model_id = "cvssp/audioldm-s-full-v2"
pipe = AudioLDMPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe = pipe.to("cuda")

print("Model loaded.")
```

## 6. Inference
Run the generation pipeline.

```
print("Generating audio...")
# num_inference_steps=10 is enough for a quick preview, 25 for better quality
audio = pipe(prompt, num_inference_steps=20, audio_length_in_s=5.0).audios[0]

# Set sample rate for saving
save_rate = 16000
print("Audio generated.")
```
