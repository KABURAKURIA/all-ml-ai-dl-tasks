## 2. Imports


```python
import torch
import pandas as pd
from transformers import AutoTokenizer, AutoModelForCausalLM
```

## 3. Load Data (Prompt Setup)
We define the system prompt to guide the model.

```python
system_prompt = "You are a helpful assistant designed to generate synthetic data for sentiment analysis. Generate 5 examples of customer reviews for a tech product. Format each example as: 'Review: <text> | Sentiment: <Positive/Negative/Neutral>'"
```

## 5. Load Model


```python
checkpoint = "HuggingFaceTB/SmolLM2-1.7B-Instruct"
device = "cuda" if torch.cuda.is_available() else "cpu"

tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModelForCausalLM.from_pretrained(checkpoint).to(device)

print("SmolLM2 loaded.")
```

## 6. Inference


```python
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": "Generate 5 examples now."}
]

input_text = tokenizer.apply_chat_template(messages, tokenize=False)
inputs = tokenizer(input_text, return_tensors="pt").to(device)

outputs = model.generate(**inputs, max_new_tokens=500, temperature=0.7)
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(generated_text)
```

## 7. Save Results
Parse and save to CSV.

```python
# Simple parsing logic (assuming model followed format)
data = []
# Split by newline and look for the pattern
lines = generated_text.split('\n')
for line in lines:
    if "|" in line and "Review:" in line:
        try:
            parts = line.split("|")
            review = parts[0].replace("Review:", "").strip()
            sentiment = parts[1].replace("Sentiment:", "").strip()
            data.append({"review": review, "sentiment": sentiment})
        except:
            continue

df = pd.DataFrame(data)
df.to_csv("synthetic_data.csv", index=False)
print(f"Saved {len(df)} examples to synthetic_data.csv")
print(df.head())
```
