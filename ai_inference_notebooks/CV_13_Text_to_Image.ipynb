## 2. Imports


```python
from diffusers import StableDiffusionPipeline
import torch
```

## 3. Load Data
Define inputs.

```python
raw_prompt = "  A futuristic city on Mars,   cinematic lighting, highly detailed  "
```

## 4. Data Cleaning
Format inputs.

```python
# We enhance the prompt to guide the model towards higher quality results.
# Adding keywords like "highly detailed", "8k", and "photorealistic" acts as a "style modifier".
enhanced_prompt = raw_prompt.strip() + ", highly detailed, 8k, photorealistic"
print(f"Enhanced Prompt: {enhanced_prompt}")
```

## 5. Load Model


```python
import torch
device = "cuda" if torch.cuda.is_available() else "cpu"
dtype = torch.float16 if device == "cuda" else torch.float32

pipe = StableDiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5", torch_dtype=dtype)
pipe = pipe.to(device)
print(f"Model loaded on {device}.")
```

## 6. Inference


```python
image = pipe(enhanced_prompt).images[0]
display(image)
```
